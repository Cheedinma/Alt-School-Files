```python
import random as r

def choose_language():
    words = ["HTML5", "Python","Java","C", "Ruby", "JavaScript"]
    return r.choice(words)

choose_language()
```

```python
def display_word(word_to_guess, guessed_word):
    display =" "
    for word in word_to_guess:
        if word in guessed_word:
            display += word
            return display
```

```python
def programming():
    word_to_guess = choose_language()
    guessed_word=[]
    max_attempts = 4 # adjust as needed
    
    print(" Welcome to Programming! you have", max_attempts," attempts to guess the language")

    while max_attempts > 0:
        current_display = display_word(word_to_guess, guessed_word)
        guess = input("Guess the Language " ).lower()
        
        if guess in guessed_word:
            print("You already guessed that language. Try again")
        
            
        guessed_word.append(guess)   
        
        if guess not in word_to_guess:
            max_attempts -= 1
            print(f" Incorrect! {max_attempts} attempts remaining")  
            continue
             
        if set (guessed_word) == (word_to_guess):
            print("Congratulations! you have guessed the language.")
            break 
            
                
    if max_attempts  == 0:
            print(f" Sorry, you ran out of attempts. The language was {word_to_guess} ")
    
```

```python
programming()
```
